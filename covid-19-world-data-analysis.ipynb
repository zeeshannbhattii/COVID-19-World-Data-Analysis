# COVID-19 World Data Analysis

Exploratory Data Analysis (EDA) using Python, Pandas, NumPy, Matplotlib, and Seaborn.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Markdown

sns.set_theme(style="whitegrid")


def section(title):
    """Print a bold section heading using Markdown."""
    display(Markdown(f"**{title}**"))
```

## Load Dataset

```python
df = pd.read_csv(
    "/kaggle/input/datasets/imdevskp/corona-virus-report/covid_19_clean_complete.csv"
)
display(df.head())
```

## Dataset Shape

```python
print("Rows, Columns:", df.shape)
```

## Column Names

```python
print(list(df.columns))
```

## Dataset Info

```python
df.info()
```

## First 5 Rows

```python
display(df.head())
```

## Last 5 Rows

```python
display(df.tail())
```

## Missing Values

```python
print(df.isnull().sum())
```

## Duplicate Rows

```python
print("Duplicates found:", df.duplicated().sum())

df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
```

## Date Column Conversion

```python
df["Date"] = pd.to_datetime(df["Date"])
print("Date dtype:", df["Date"].dtype)
```

## Data Types

```python
print(df.dtypes)
```

## Statistical Summary

```python
display(df.describe())
```

## Confirmed Cases Stats

```python
print(f"Average : {df['Confirmed'].mean():,.2f}")
print(f"Minimum : {df['Confirmed'].min():,.0f}")
print(f"Maximum : {df['Confirmed'].max():,.0f}")
```

## Deaths Stats

```python
print(f"Average : {df['Deaths'].mean():,.2f}")
print(f"Minimum : {df['Deaths'].min():,.0f}")
print(f"Maximum : {df['Deaths'].max():,.0f}")
```

## Recovered Stats

```python
print(f"Average : {df['Recovered'].mean():,.2f}")
print(f"Minimum : {df['Recovered'].min():,.0f}")
print(f"Maximum : {df['Recovered'].max():,.0f}")
```

## Number Of Countries

```python
print("Total countries:", df["Country/Region"].nunique())
```

## Country List

```python
print(df["Country/Region"].unique())

country_cases = df.groupby("Country/Region")["Confirmed"].sum()
top_cases = country_cases.sort_values(ascending=False).head(10)
```

## Top 10 Countries - Confirmed Cases

```python
print(top_cases.to_string())

plt.figure(figsize=(10, 5))
plt.bar(top_cases.index, top_cases.values, color="steelblue")
plt.title("Top 10 Countries by Confirmed COVID-19 Cases", fontsize=13, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

country_deaths = df.groupby("Country/Region")["Deaths"].sum()
top_deaths = country_deaths.sort_values(ascending=False).head(10)
```

## Top 10 Countries - Deaths

```python
print(top_deaths.to_string())

plt.figure(figsize=(10, 5))
plt.bar(top_deaths.index, top_deaths.values, color="firebrick")
plt.title("Top 10 Countries by COVID-19 Deaths", fontsize=13, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Deaths")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

country_recovered = df.groupby("Country/Region")["Recovered"].sum()
top_recovered = country_recovered.sort_values(ascending=False).head(10)
```

## Top 10 Countries - Recovered

```python
print(top_recovered.to_string())

plt.figure(figsize=(10, 5))
plt.bar(top_recovered.index, top_recovered.values, color="seagreen")
plt.title("Top 10 Countries by Recovered Cases", fontsize=13, fontweight="bold")
plt.xlabel("Country")
plt.ylabel("Recovered Cases")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Time-Series — Cases Over Time

```python
daily_cases = df.groupby("Date")["Confirmed"].sum()
plt.figure(figsize=(12, 5))
plt.plot(daily_cases.index, daily_cases.values, color="steelblue", linewidth=2)
plt.title("COVID-19 Confirmed Cases Over Time", fontsize=13, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Confirmed Cases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Time-Series — Deaths Over Time

```python
daily_deaths = df.groupby("Date")["Deaths"].sum()
plt.figure(figsize=(12, 5))
plt.plot(daily_deaths.index, daily_deaths.values, color="firebrick", linewidth=2)
plt.title("COVID-19 Deaths Over Time", fontsize=13, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Deaths")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Time-Series — Recoveries Over Time

```python
daily_recovered = df.groupby("Date")["Recovered"].sum()
plt.figure(figsize=(12, 5))
plt.plot(daily_recovered.index, daily_recovered.values, color="seagreen", linewidth=2)
plt.title("COVID-19 Recoveries Over Time", fontsize=13, fontweight="bold")
plt.xlabel("Date")
plt.ylabel("Recovered Cases")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## Distribution Of Confirmed Cases

```python
plt.figure(figsize=(10, 5))
sns.barplot(x=top_cases.values, y=top_cases.index, hue=top_cases.index, palette="Blues_r", legend=False)
plt.title("Distribution of Confirmed COVID-19 Cases", fontsize=13, fontweight="bold")
plt.xlabel("Confirmed Cases")
plt.tight_layout()
plt.show()
```

## Seaborn - Top 10 Countries

```python
plt.figure(figsize=(10, 5))
sns.barplot(x=top_cases.values, y=top_cases.index, hue=top_cases.index, palette="Blues_r", legend=False)
plt.title("Top 10 Countries by Confirmed Cases", fontsize=13, fontweight="bold")
plt.xlabel("Confirmed Cases")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```

## Correlation Matrix

```python
correlation = df[["Confirmed", "Deaths", "Recovered"]].corr()
display(correlation)

plt.figure(figsize=(8, 5))
sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("COVID-19 Correlation Heatmap", fontsize=13, fontweight="bold")
plt.tight_layout()
plt.show()
```

## Numpy - Confirmed Cases Stats

```python
confirmed = np.array(df["Confirmed"].dropna())
print(f"Average             : {np.mean(confirmed):,.2f}")
print(f"Maximum             : {np.max(confirmed):,.0f}")
print(f"Minimum             : {np.min(confirmed):,.0f}")
print(f"Standard Deviation  : {np.std(confirmed):,.2f}")
```

## Highest Confirmed Case Record

```python
print(df.loc[df["Confirmed"].idxmax()])
```

## Highest Death Record

```python
print(df.loc[df["Deaths"].idxmax()])
```

## Overall Comparison

```python
comparison = pd.Series({
    "Confirmed": df["Confirmed"].sum(),
    "Deaths": df["Deaths"].sum(),
    "Recovered": df["Recovered"].sum()
})
print(comparison.to_string())

plt.figure(figsize=(8, 5))
sns.barplot(x=comparison.index, y=comparison.values, hue=comparison.index,
            palette=["steelblue", "firebrick", "seagreen"], legend=False)
plt.title("COVID-19 Cases Comparison", fontsize=13, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Total")
plt.tight_layout()
plt.show()
```

## Top 10 Countries - Latest Date (Corrected)

```python
latest_date = df["Date"].max()
latest_data = df[df["Date"] == latest_date]
top_latest = latest_data.sort_values("Confirmed", ascending=False).head(10)

print("Latest date in dataset:", latest_date.date())
display(top_latest[["Country/Region", "Confirmed", "Deaths", "Recovered"]])

plt.figure(figsize=(10, 5))
sns.barplot(data=top_latest, x="Confirmed", y="Country/Region",
            hue="Country/Region", palette="Blues_r", legend=False)
plt.title("Top 10 Countries by Confirmed Cases (Latest Date)", fontsize=13, fontweight="bold")
plt.xlabel("Confirmed Cases")
plt.ylabel("Country")
plt.tight_layout()
plt.show()
```
